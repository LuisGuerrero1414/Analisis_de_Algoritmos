$$T_1(n) \le c \cdot T_2(n)$$

### **1. Funciones de Piso ($\lfloor x \rfloor$) y Techo ($\lceil x \rceil$) en Computación**

En matemáticas discretas y programación, no siempre podemos trabajar con números decimales (por ejemplo, no puedes tener "medio" ciclo de un bucle o "3.5" bits).

* **Piso ($\lfloor x \rfloor$):** Redondea cualquier número real hacia abajo al entero más cercano.
* **Techo ($\lceil x \rceil$):** Redondea cualquier número real hacia arriba al entero más cercano.

#### **¿Para qué las usamos en nuestra carrera?**

* **Cálculo de Bits:** Para saber cuántos bits necesitas en memoria para codificar y diferenciar $n$ objetos, la fórmula exacta es el techo del logaritmo: $\lceil \log_2 n \rceil$. Por ejemplo, si tienes 1000 elementos, $\log_2(1000) \approx 9.96$, por lo que necesitas $\lceil 9.96 \rceil = 10$ bits.
* **Dividir problemas a la mitad (Divide y Vencerás):** Cuando divides un arreglo de tamaño impar $n$ en dos partes, usas la propiedad $\lceil n/2 \rceil + \lfloor n/2 \rfloor = n$ para asegurar que una mitad tome la parte superior, la otra la inferior, y ningún dato se pierda en el camino.
* **Aritmética de residuos (Operador módulo):** Para calcular el residuo de una división entera en el procesador de forma matemática, la operación se define usando la función piso: $a \text{ mod } n = a - n \lfloor a/n \rfloor$.

### **2. ¿Por qué la función exponencial crece más rápido que cualquier polinomio?**

Como ingeniero en sistemas, debes evitar a toda costa los algoritmos con crecimiento exponencial, y la matemática nos explica claramente por qué:

#### **La razón matemática (El Límite)**

Para cualquier base constante de una exponencial $a > 1$ (como $2^n$ o $3^n$) y cualquier exponente constante de un polinomio $b$ (como $n^2$ o $n^{100}$), se cumple el siguiente límite asintótico cuando los datos ($n$) tienden al infinito:

$$\lim_{n \to \infty} \frac{n^b}{a^n} = 0$$

Esto demuestra que $n^b = o(a^n)$ (notación o pequeña), lo que significa que la función exponencial crece infinitamente más rápido y domina por completo a cualquier polinomio conforme $n$ se hace grande.

#### **¿Qué impacto tiene esto en el software real?**

Esta diferencia matemática separa lo tratable (lo que una computadora puede resolver en un tiempo razonable) de lo intratable:

* Un algoritmo con complejidad polinomial cuadrática ($n^2$) procesando $n = 100$ datos toma solo 10,000 operaciones.
* Un algoritmo exponencial ($2^n$) con los mismos $n = 100$ datos requiere $1.26 \times 10^{30}$ operaciones. ¡Eso tardaría miles de millones de años en procesarse, incluso en la supercomputadora más rápida del mundo!

### **¿Qué es un algoritmo y cuáles son sus características clave?**

Para ti que estás iniciando tu formación en Ingeniería en Sistemas Computacionales, un algoritmo es, de manera intuitiva, una serie finita de pasos diseñados para resolver un problema. Formalmente, se define como *"un conjunto de pasos que, ejecutados de la manera correcta, permiten obtener un resultado (en un tiempo acotado)"*, constituyendo un método sistemático, con reglas operacionales concretas, precisas y no ambiguas que pueden realizarse mecánicamente por una computadora para transformar datos de entrada en datos de salida.

Para que un procedimiento sea considerado un verdadero algoritmo, debe cumplir rigurosamente con estas características clave:

* **Entrada y Salida:** Debe tener definidos de forma clara los datos iniciales que requiere (entrada) y el resultado o solución específica que produce (salida).
* **No ambigüedad (Precisión):** Debe indicar explícitamente y sin dudas el orden exacto de los pasos a seguir de modo que incluso una máquina pueda ejecutarlos.
* **Finitud:** Debe terminar obligatoriamente después de un número finito de pasos. Si un procedimiento es correcto pero no tiene un final, se le conoce simplemente como "método computacional".
* **Determinismo:** Si ejecutas el algoritmo varias veces utilizando el mismo conjunto de datos de entrada, siempre debe pasar por los mismos estados y producir exactamente el mismo resultado.
* **Efectividad:** Cada instrucción debe ser lo suficientemente básica como para poder realizarse en un tiempo finito, logrando el efecto deseado.
* **Generalidad:** Debe ser capaz de contemplar y resolver correctamente todos los casos posibles de entrada.

---

### **El "Problema de Paro" de Alan Turing**

Formulado por el matemático Alan Turing en 1936, el Problema de Paro (*halting problem*) consiste en determinar si un programa de computadora arbitrario, al trabajar con una entrada dada, llegará en algún momento a detenerse o si continuará ejecutándose indefinidamente en un bucle infinito.

Turing demostró de manera matemática la insolubilidad (o indecidibilidad) de este problema. Probó que es imposible construir un algoritmo general único que sea capaz de predecir correctamente para cualquier par (programa, entrada) si este va a parar o no.

¿Por qué importa esto a un ingeniero en sistemas? Porque define los límites científicos de nuestra disciplina, demostrando que existen problemas perfectamente planteados para los cuales nunca existirá una solución algorítmica.

---

### **Análisis a Priori vs. Prueba a Posteriori**

Para medir la eficiencia de un algoritmo (el uso de recursos de tiempo y memoria), la ciencia de la computación divide este estudio en dos fases o aproximaciones claras:

#### **Análisis a Priori (Estudio Teórico):**
* **Cuándo se hace:** En la etapa de diseño, directamente sobre papel o pseudocódigo, antes de escribir una sola línea de código ejecutable.
* **Qué produce:** Una función matemática (como la notación Big O) que acota o limita el tiempo de ejecución estimado en función del tamaño de entrada $n$.
* **De qué depende:** Es totalmente independiente del hardware, del lenguaje de programación, del compilador o de tu habilidad para programar. Te permite comparar la eficiencia de dos algoritmos de forma pura y científica.

#### **Prueba a Posteriori (Estudio Experimental):**
* **Cuándo se hace:** Después de programar e implementar el algoritmo en un código ejecutable real.
* **Qué produce:** Estadísticas reales y medidas físicas del tiempo de ejecución y el espacio en memoria consumidos mientras el programa corre.
* **De qué depende:** Es altamente dependiente del entorno tecnológico, lo que significa que tus resultados cambiarán según la computadora específica que uses, el compilador, el lenguaje de programación y los datos de prueba particulares que selecciones para la ejecución.

- *Explica en qué consisten el peor caso, mejor caso y caso promedio al momento de evaluar el comportamiento de un algoritmo ante distintas entradas.*  
**RESPUESTA:**

### **1. El Peor Caso (Worst Case o $T_{\max}(n)$)**

* **¿En qué consiste?:** Es el escenario en el que el algoritmo consume el tiempo máximo, la mayor cantidad de instrucciones o el mayor número de operaciones elementales para resolver un problema de tamaño $n$.
* **Definición matemática:** Si $D_n$ es el conjunto de todas las entradas posibles de tamaño $n$ e $I$ es una entrada particular que requiere $t(I)$ operaciones elementales, el peor caso se define formalmente como el valor máximo de ese conjunto:

$$W(n) = \max \{t(I) \mid I \in D_n\}$$

* **¿Por qué es el más importante?:** Porque proporciona una cota superior (un límite máximo) garantizada. Esto le asegura al ingeniero que, bajo cualquier circunstancia o combinación de datos de entrada, el programa nunca tardará más que el límite calculado.

---

### **2. El Mejor Caso (Best Case o $T_{\min}(n)$)**

* **¿En qué consiste?:** Es el escenario en el que el algoritmo se ejecuta de la forma más rápida y óptima posible, consumiendo el tiempo mínimo o requiriendo la menor cantidad de operaciones elementales para resolver el problema.
* **Ejemplo clásico:** En una búsqueda secuencial dentro de un arreglo de tamaño $n$, el mejor caso ocurre cuando el elemento que estamos buscando se encuentra en la primerísima posición del arreglo, permitiendo al algoritmo terminar con éxito tras realizar una sola comparación.
* **Su limitación:** Evaluar un algoritmo únicamente por su mejor caso es engañoso. Al representar una situación sumamente optimista que rara vez ocurre en condiciones reales, no proporciona información verdaderamente útil sobre el desempeño típico del software.

---

### **3. El Caso Promedio (Average Case)**

* **¿En qué consiste?:** Es el tiempo esperado o rendimiento medio del algoritmo al resolver un problema cualquiera de tamaño $n$, considerando todas las entradas posibles.
* **¿Por qué es valioso?:** Representa la medida de rendimiento cotidiana más realista y típica de un programa informático en producción.
* **Su dificultad matemática:** Es el análisis más difícil de calcular con diferencia. Para obtenerlo, no basta de ninguna manera con promediar los resultados del peor y el mejor caso; es obligatorio conocer o asumir una distribución de probabilidad estadística de los datos de entrada en el mundo real. Habitualmente, para simplificar los análisis matemáticos, se asume la hipótesis de que todas las entradas de tamaño $n$ tienen la misma probabilidad de presentarse, aunque esto pueda o no ser cierto en la práctica.

````python
def fibonacci_iterativo(n):
    if n <= 1:
        return n
    a, b = 0, 1
    for _ in range(2, n + 1):
        a, b = b, a + b
    return b

# Ejemplo de uso:
n = 6
print(f"El número de Fibonacci en la posición {n} (iterativo) es: {fibonacci_iterativo(n)}")
````

2. Utiliza la librería `time` o la función mágica `%timeit` para comparar el tiempo consumido (A posteriori) entre el Fibonacci recursivo y el iterativo para `n=35`.

````python
import time

# --- Algoritmo Iterativo ---
def fibonacci_iterativo(n):
    if n == 0:
        return 0
    if n == 1:
        return 1

    a, b = 0, 1
    for _ in range(2, n + 1):
        a, b = b, a + b
    return b

# --- Algoritmo Recursivo ---
def fibonacci_recursivo(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return fibonacci_recursivo(n - 1) + fibonacci_recursivo(n - 2)

# --- Medición de Tiempos ---
n = 35

# Medición Versión Iterativa
inicio_iter = time.perf_counter()
res_iter = fibonacci_iterativo(n)
fin_iter = time.perf_counter()
tiempo_iter = fin_iter - inicio_iter

# Medición Versión Recursiva
inicio_rec = time.perf_counter()
res_rec = fibonacci_recursivo(n)
fin_rec = time.perf_counter()
tiempo_rec = fin_rec - inicio_rec

# --- Impresión de Resultados ---
print(f"Fibonacci({n}) = {res_iter}\n")
print(f"Tiempo Iterativo : {tiempo_iter:.8f} segundos")
print(f"Tiempo Recursivo : {tiempo_rec:.8f} segundos")
print(f"\nEl enfoque iterativo fue ~{tiempo_rec / tiempo_iter:,.0f} veces más rápido.")
````
